```python
import numpy as np
import pandas as pd
import torch
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/gscdit/Breast-Cancer-Detection/refs/heads/master/data.csv')
df.head()
```

```python
df.shape
```

```python

df.drop(columns=['id', 'Unnamed: 32'], inplace= True)
```

```python
df.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(df.iloc[:, 1:], df.iloc[:, 0], test_size=0.2)
```

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
encoder = LabelEncoder()
y_train = encoder.fit_transform(y_train)
y_test = encoder.transform(y_test)
```

```python
X_train_ten=torch.from_numpy(X_train)
X_test_ten=torch.from_numpy(X_test)
y_train_ten=torch.from_numpy(y_train)
y_test_ten=torch.from_numpy(y_test)
```

```python
print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("X_train_ten:", X_train_ten.shape)
print("y_train_ten:", y_train_ten.shape)
```

```python
X_train_ten = X_train_ten.float()
X_test_ten = X_test_ten.float()

y_train_ten = y_train_ten.float()
y_test_ten = y_test_ten.float()
```

```python
from torch.utils.data import DataLoader, Dataset

class CustomDataset(Dataset):

  def __init__(self,features,labels):

    self.features=features
    self.labels=labels

  def __len__(self):
    return len(self.features)

  def __getitem__(self,index):

    return self.features[index],self.labels[index]
```

```python
train_dataset=CustomDataset(X_train_ten,y_train_ten)
test_dataset=CustomDataset(X_test_ten,y_test_ten)
```

```python
train_dataset.__getitem__(10)
```

```python
train_loader=DataLoader(train_dataset,batch_size=32,shuffle=True)
test_loader=DataLoader(test_dataset,batch_size=32,shuffle=True)
```

```python
#Defining the NN
```

```python
import torch
import torch.nn as nn

# Defining Neural Network
class SimpleNN(nn.Module):

  def __init__(self,num_features):

   super().__init__()
   self.linear=nn.Linear(num_features,1)
   self.sigmoid=nn.Sigmoid()

  def forward(self,features):

   out=self.linear(features)
   out=self.sigmoid(out)
   return out
```

```python
learning_rate=0.1
epochs=25
```

```python
loss_function=nn.BCELoss()
```

```python
model=SimpleNN(X_train_ten.shape[1])
model.linear.weight
```

```python
model.linear.bias
```

```python
#Training

optimizer=torch.optim.SGD(model.parameters(),lr=learning_rate)
for epoch in range(epochs):

  for batch_features,batch_labels in train_loader:
    # Forward Propagation
    y_pred = model(batch_features)

    # Loss Calculation
    loss = loss_function(y_pred, batch_labels.view(-1, 1))

     # Reset Gradients
    optimizer.zero_grad()
    # Backpropagation
    loss.backward()

    # Gradient Descent
    optimizer.step()


    print(f'Epoch: {epoch+1}, Loss: {loss.item():.4f}')
```

```python
#evaluation

with torch.no_grad():
  y_pred=model.forward(X_test_ten)
  y_pred=(y_pred > 0.7).float()
  accuracy=(y_pred==y_test_ten).float().mean()



  print(y_pred)
```

```python
  print(f'Accuracy : {accuracy.item()}')
```
